```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

reviews = [
    "This product is excellent",
    "I love this product",
    "Very good quality",
    "Amazing experience",
    "This product is bad",
    "Very poor quality",
    "I hate this product",
    "Worst experience"
]

labels = np.array([1, 1, 1, 1, 0, 0, 0, 0])

tokenizer = Tokenizer(num_words=1000, oov_token="<OOV>")
tokenizer.fit_on_texts(reviews)

X = tokenizer.texts_to_sequences(reviews)
X = pad_sequences(X, maxlen=6)

model = Sequential([
    Embedding(1000, 16),
    SimpleRNN(32),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(X, labels, epochs=30, verbose=1)

test = ["This product is very good"]
test = pad_sequences(tokenizer.texts_to_sequences(test), maxlen=6)

prediction = model.predict(test, verbose=0)

print("Positive Review" if prediction[0][0] >= 0.5
      else "Negative Review")
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

text = "deep learning is very interesting"

chars = sorted(set(text))
char_to_id = {c: i for i, c in enumerate(chars)}
id_to_char = {i: c for c, i in char_to_id.items()}

X, y = [], []

for i in range(len(text)-5):
    X.append([char_to_id[c] for c in text[i:i+5]])
    y.append(char_to_id[text[i+5]])

X = np.array(X)
y = np.array(y)

model = Sequential([
    SimpleRNN(64, input_shape=(5, 1)),
    Dense(len(chars), activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy')

model.fit(X.reshape(-1, 5, 1), y, epochs=100, verbose=0)

seed = "deep "
generated = seed

for i in range(20):
    seq = [char_to_id[c] for c in generated[-5:]]
    pred = model.predict(np.array(seq).reshape(1, 5, 1), verbose=0)
    next_char = id_to_char[np.argmax(pred)]
    generated += next_char

print("Generated Text:", generated)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from sklearn.metrics import mean_squared_error

data = np.sin(np.arange(200) * 0.1) + 10

X, y = [], []

for i in range(10, len(data)):
    X.append(data[i-10:i])
    y.append(data[i])

X = np.array(X).reshape(-1, 10, 1)
y = np.array(y)

split = int(len(X) * 0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

model = Sequential([
    SimpleRNN(32, input_shape=(10, 1)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y_train, epochs=10, verbose=0)

pred = model.predict(X_test, verbose=0)

print("Actual:", y_test[:5])
print("Predicted:", pred[:5].flatten())
print("MSE:", mean_squared_error(y_test, pred))
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

text = "deep learning is interesting deep learning is useful"

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

seq = tokenizer.texts_to_sequences([text])[0]

X, y = [], []

for i in range(1, len(seq)):
    X.append(seq[:i])
    y.append(seq[i])

X = pad_sequences(X)
y = np.array(y)

model = Sequential([
    Embedding(len(tokenizer.word_index)+1, 16),
    SimpleRNN(32),
    Dense(len(tokenizer.word_index)+1, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=0)

test = pad_sequences(
    [tokenizer.texts_to_sequences(["deep learning"])[0]],
    maxlen=X.shape[1]
)

pred = model.predict(test, verbose=0)
word_id = np.argmax(pred)

print("Next Word:", tokenizer.index_word.get(word_id, "Unknown"))
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, Dense

data = np.sin(np.arange(200) * 0.1)

X, y = [], []

for i in range(10, len(data)):
    X.append(data[i-10:i])
    y.append(data[i])

X = np.array(X).reshape(-1, 10, 1)
y = np.array(y)

def create_model(layer):
    model = Sequential([
        layer(32, input_shape=(10, 1)),
        Dense(1)
    ])
    model.compile(optimizer='adam', loss='mse')
    return model

rnn = create_model(SimpleRNN)
lstm = create_model(LSTM)

rnn.fit(X, y, epochs=10, verbose=0)
lstm.fit(X, y, epochs=10, verbose=0)

print("RNN Loss:", rnn.evaluate(X, y, verbose=0))
print("LSTM Loss:", lstm.evaluate(X, y, verbose=0))
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

data = np.sin(np.arange(200) * 0.1)

X, y = [], []

for i in range(10, len(data)):
    X.append(data[i-10:i])
    y.append(data[i])

X = np.array(X).reshape(-1, 10, 1)
y = np.array(y)

for units in [16, 32]:
    for lr in [0.001, 0.01]:
        for batch in [16, 32]:

            model = Sequential([
                LSTM(units, input_shape=(10, 1)),
                Dense(1)
            ])

            model.compile(
                optimizer=__import__('tensorflow').keras.optimizers.Adam(
                    learning_rate=lr
                ),
                loss='mse'
            )

            model.fit(X, y, epochs=5,
                      batch_size=batch, verbose=0)

            loss = model.evaluate(X, y, verbose=0)

            print("Units:", units,
                  "Learning Rate:", lr,
                  "Batch Size:", batch,
                  "Loss:", loss)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from sklearn.metrics import precision_score, recall_score, f1_score

X = np.random.randint(1, 50, (100, 10))
y = np.random.randint(0, 3, 100)

model = Sequential([
    Embedding(50, 16),
    SimpleRNN(32),
    Dense(3, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.fit(X, y, epochs=5, verbose=0)

pred = np.argmax(model.predict(X, verbose=0), axis=1)

print("Precision:", precision_score(y, pred, average='macro',
                                    zero_division=0))
print("Recall:", recall_score(y, pred, average='macro',
                              zero_division=0))
print("F1 Score:", f1_score(y, pred, average='macro',
                            zero_division=0))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Vocabulary
words = ["good", "bad", "movie", "excellent", "poor"]

word_id = {w: i for i, w in enumerate(words)}

class RecursiveNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.embedding = nn.Embedding(5, 16)
        self.combine = nn.Linear(32, 16)
        self.output = nn.Linear(16, 2)

    def forward(self, tree):
        if isinstance(tree, str):
            return self.embedding(
                torch.tensor(word_id[tree])
            )

        left = self.forward(tree[0])
        right = self.forward(tree[1])

        combined = torch.cat((left, right))
        return torch.tanh(self.combine(combined))

    def predict(self, tree):
        result = self.forward(tree)
        return self.output(result)

model = RecursiveNN()
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss()

# Tree-structured training data
data = [
    (("good", "movie"), 1),
    (("excellent", "movie"), 1),
    (("bad", "movie"), 0),
    (("poor", "movie"), 0)
]

# Training
for epoch in range(100):
    for tree, label in data:
        optimizer.zero_grad()

        output = model.predict(tree)
        target = torch.tensor([label])

        loss = criterion(output.unsqueeze(0), target)
        loss.backward()
        optimizer.step()

# Testing
test = ("good", "movie")
prediction = model.predict(test)

result = torch.argmax(prediction).item()

print("Positive Sentiment" if result == 1
      else "Negative Sentiment")
```
